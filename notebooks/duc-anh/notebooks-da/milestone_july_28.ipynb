# Milestone July 28: Understand & Inspect
This notebook dynamically locates and previews the raw datasets before validation.


```python
import pandas as pd
import glob
import os

print("=== MILESTONE JULY 28: RAW DATA INSPECTION ===")
raw_dir = '../data_raw'
files = glob.glob(os.path.join(raw_dir, '**', '*.csv*'), recursive=True)

if len(files) > 0:
    print(f" Found {len(files)} raw data partitions.")
    print(f"-> Previewing sample from: {files[0]}")
    try:
        sample_df = pd.read_csv(files[0], compression='gzip' if files[0].endswith('.gz') else None, nrows=5)
        display(sample_df)
    except Exception as e:
        print(f"Error reading file: {e}")
else:
    print(" WARNING: No raw data found in '../data_raw'. Please check directory structure.")

```
